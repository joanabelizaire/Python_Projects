```python
# Convert the date column to datetime format (if not already)
df['Created Date'] = pd.to_datetime(df['Created Date'])

# Find the earliest and latest date
print("Earliest date:", df['Created Date'].min())
print("Latest date:", df['Created Date'].max())

```

                                  Question 3

```python
# Number of unique complaint types
# This shows you how many different complaint types there are — such as 'Food Poisoning', 'Noise - Residential', and so on."

print("Unique complaint types:", df['Complaint Type'].nunique())
```

```python
# Most common complaint type
# mode() finds the value that appears most frequently in the column (the same thing as "top" in describe()).
# The [0] is there because .mode() can return multiple values if there's a tie — we just take the first one.

print("Most common complaint:", df['Complaint Type'].mode()[0])
```

```python
# Number of unique boroughs
# This shows how many different boroughs there are in the dataset (Brooklyn, Manhattan, Queens, Bronx, Staten Island → should give 5).

print("Unique boroughs:", df['Borough'].nunique())
```

```python
# Show all unique borough values
print(df['Borough'].unique())
```

```python
print(df['Borough'].value_counts())
```

```python
# Most common agency handling complaints
# This shows which agency (DOHMH, NYPD, DEP, etc.) handles the most complaints in the dataset.

print("Most common agency:", df['Agency'].mode()[0])
```

                                     Question 4

```python
import matplotlib.pyplot as plt

# Visualization 1: Top 10 most common complaint types

df['Complaint Type'].value_counts().head(10).plot(kind='bar', title='Top 10 Complaint Types')
plt.xlabel('Complaint Type')
plt.ylabel('Number of Complaints')
plt.show()
```

```python
# Visualization 2: Number of complaints by borough

df['Borough'].value_counts().plot(kind='bar', title='Complaints by Borough', color='orange')
plt.xlabel('Borough')
plt.ylabel('Number of Complaints')
plt.show()
```

```python
# top 7 complaints
top_complaints = df["Complaint Type"].value_counts().head(7)
top_complaints.plot(kind="pie")
plt.title("Top 7 Complaints")
plt.xlabel("Complaint type")
plt.ylabel("Number of complaints")
plt.show()
```

```python
# Visualization 3: Number of complaints by agency
df['Agency'].value_counts().head(10).plot(kind='bar', title='Top 10 Agencies by Complaint Count', color='green')
plt.xlabel('Agency')
plt.ylabel('Number of Complaints')
plt.show()
```

                                  Question 6

```python
# Filter the dataset to only keep rows where Complaint Type contains "Noise"
noise_complaints = df[df['Complaint Type'].str.contains('Noise', na=False)]

# Count how many noise complaints there are per borough, and plot it

noise_complaints['Borough'].value_counts().plot(kind='bar', title='Noise Complaints by Borough', color='purple')
plt.xlabel('Borough')
plt.ylabel('Number of Noise Complaints')
plt.show()
```

                                   Question 7

```python
# Convert both date columns to datetime

df['Created Date'] = pd.to_datetime(df['Created Date'])
df['Closed Date'] = pd.to_datetime(df['Closed Date'])

# Calculate how long each complaint took to resolve
df['Resolution Time'] = df['Closed Date'] - df['Created Date']

# Average resolution time
print("Average resolution time:", df['Resolution Time'].mean())
```

```python
# Extract the day of the week from the Created Date
df['Day of Week'] = df['Created Date'].dt.day_name()

# Count complaints by day of week
print(df['Day of Week'].value_counts())
```

```python
# Extract the hour from the Created Date
df['Hour'] = df['Created Date'].dt.hour

# Count complaints by hour and plot
df['Hour'].value_counts().sort_index().plot(kind='bar', title='Complaints by Hour of Day')
plt.xlabel('Hour of Day')
plt.ylabel('Number of Complaints')
plt.show()
```
