##           Clean the dataset by naming the missing column name

```python
# Rename the column with the missing/unnamed name to something meaningful
df = df.rename(columns={'Unnamed: 0': 'Car_model'})
```

```python
# Verify the change worked
df.head()
```

```python
df.columns
```

Create a new dataframe that consists of a subset of the original
 columns (only mpg, hp, wt, and cyl)

```python
# Create a new dataframe with only mpg, hp, wt, and cyl columns
df2 = df[['mpg', 'hp', 'wt', 'cyl']]
```

```python
# Verify the result
df2
```

##  Rename the above columns (cylinders = cyl)

```python
# Rename the 'Cyl' column to 'Cylinder'
df2 = df2.rename(columns={'cyl': 'cylinders'})
```

```python
# Verify the change
df2.head()
```

## Summary statistics for those columns (average mpg)

```python
# Summary statistics for the columns (including average mpg)
df2.describe()
```

```python
# just the average mpg
df2['mpg'].mean()
```
